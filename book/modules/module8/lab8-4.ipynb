```python
%matplotlib inline
import numpy as np
import scipy
import matplotlib.pyplot as plt
```

```python
#create a very simple matrix that can be easily explained
X = np.array([
    [2, 1, -1, -2, 0],  
    [2, 1, -1, -2, 0],
    [-1, -0.5, 0.5, 1, 0]
])    

Y = np.array([
    [10, 12, 8, 15, 11],   # Precipitation
    [20, 24, 17, 30, 22],  # Snow depth
    [15, 18, 12, 25, 16]   # Runoff
])
```

```python
np.set_printoptions(precision=2, suppress=True)
print(X)
```

```python
# Calculate covariance matrix
C = np.cov(X)

print(C)

C2 = np.cov(X, rowvar=False)

print(C2)
```

```python
U, S, Vt = np.linalg.svd(X, full_matrices=False)

print("U:", U)
print("Singular values:", S)
print("V transpose:", Vt)
```

```python
variance_explained = S**2 / np.sum(S**2)

print(variance_explained)
print("Cumulative:", np.cumsum(variance_explained))
```

## What we see here is that all the variance in X is explained by one mode, the first EOF.  Essentially all of these are on one line
Also, you see that U is a 3x3 matrix, S is a 3x1 matrix, and V transpose is a 5x3 matrix

```python
print("U1 = ", U[:,0])
print("S1 = ",S[0])
print("Vt1 = ", Vt[0,:])
```

```python
#And we can then reconstruct our temperature matrix from just this
```

```python
# First EOF (spatial pattern)
EOF1 = U[:, 0]

# First PC (temporal pattern)
PC1 = Vt[0, :]

# First singular value
S1 = S[0]

# Reconstruct matrix using only first mode
X_reconstructed = S1 * np.outer(EOF1, PC1)

print(X_reconstructed)
```
